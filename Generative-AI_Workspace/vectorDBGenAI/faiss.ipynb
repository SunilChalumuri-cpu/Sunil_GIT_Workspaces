```python
import requests 
import json
import os 
import numpy as np
# Broken down the text into smaller chunks
def chunk_text(text, chunk_size=50, overlap=5):
    words = text.split()
    chunks = []
    start = 0
    while start < len(words):
        end = min(start + chunk_size, len(words))
        chunks.append(" ".join(words[start:end]))
        start += chunk_size - overlap
    return chunks
```

```python
chunks = chunk_text(bio_text)
print(f"Total chunks created: {len(chunks)}")
chunks
```

```python
API_URL = "https://api.euron.one/api/v1/euri/embeddings"
API_KEY = "euri-eca149f81167bfd14dcbf6025b8bb834fec9aa6b728b4e8bac19531047ecb9f0"
API_MODEL = "text-embedding-3-small"
headers = {
    "Content-Type": "application/json",
    "Authorization": f"Bearer {API_KEY}"
}
```

```python
# Generate embeddings for each chunk of text
import urllib3
urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

all_embeddings = []
for i, chunk in enumerate(chunks):
    payload = {
        "model": API_MODEL,
        "input": chunk
    }
    response = requests.post(API_URL, headers=headers, data=json.dumps(payload), verify=False)
    result = response.json()
    all_embeddings.append(result["data"][0]["embedding"])
```

```python
# Convert the list of embeddings to a NumPy array
embeddings_array =np.array(all_embeddings, dtype=np.float32)
embeddings_array

```

```python
# Create a FAISS index of same dimensionality(1536 here) and add the embeddings
# We use IndexFlatL2 for simplicity, which performs exact nearest neighbor search using L2 distance.
base_index = faiss.IndexFlatL2(embeddings_array.shape[1])
base_index.add(embeddings_array)
```

```python
# Save the FAISS index to disk
faiss.write_index(base_index, "faiss_index.faiss")
```

```python
# Querying the FAISS index
query_text = "What is Sudhanshu's background?"
```

```python
# Function to get embedding for the query
def get_query_embedding(query):
    payload = {
        "model": API_MODEL,
        "input": query
    }
    response = requests.post(API_URL, headers=headers, data=json.dumps(payload), verify=False)
    result = response.json()
    return np.array(result["data"][0]["embedding"], dtype=np.float32)
```

```python
# Calling the function to get the query embedding
query_embedding = get_query_embedding(query_text)
query_embedding
```

```python
# Performing the search in the FAISS index for the top 3 most similar chunks
base_index.search(np.array([query_embedding]), k=3)
```
