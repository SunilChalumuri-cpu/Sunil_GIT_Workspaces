```python
def decorator_custome(func):
    def impl(a,b):
        print(f"The function {func.__name__} is called with {a} and {b}")
        result = func(a,b)
        print(f"The result of {func.__name__} is {result}")
        return result
    return impl
```

```python
@decorator_custome
def sub(a,b):
    c = a-b
    return c
```

```python
sub(4,5)
```

```python
@decorator_custome
def add(a,b):
    
    c = a+b
    
    return c
```

```python
import time
def loop_list(l):
    start_time = time.time()
    for i in l:
        time.sleep(2)
        print(i)
    
    end_time = time.time()
    print(f"The execution time is {end_time - start_time}")
```

```python
loop_list([3,4,5,6,7,87,8,89,9,5])
```

```python
def time_calculator(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"The execution time of {func.__name__} is {end_time - start_time} seconds")
        return result
    return wrapper
```

```python
@time_calculator
def loop_list1(l):
    for i in l:
        time.sleep(1)
        print(i)
```

```python
loop_list1([2,3,4,5,6,7,88,89])
```

```python
from functools import lru_cache

@lru_cache(maxsize=500)
def fibonacci(n):
    if n<=1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)
```

```python
fibonacci(5)
```

```python
fibonacci(10)
```

```python
fibonacci(50)
```

```python
data = "+91 34563535646 , +1 3453636354, +45 23535346 , asdfsffas,f asdffafas,fas,fa,fas,fas,fa, sudhanshu@euron.one"

# Phone Number Pattern
phone_pattern = r'\+?\d{1,4}?[\s.-]?\(?\d{3}\)?[\s.-]?\d{3}[\s.-]?\d{4}|\+?\d{1,2}[\s.-]?\d{3,5}[\s.-]?\d{3,5}|\d{3}[-\.\ ]?\d{3}[-\.\ ]?\d{4}|[0-9]{2,3}-[0-9]{2,3}-[0-9]{2,3}|[0-9]{3}-[0-9]{3}-[0-9]{4}|[0-9]{10}|[0-9]{12}'

# Email ID Pattern
email_pattern =r"[\w\.-@]+@[a-zA-Z0-9.-]+\.[a-zA-Z]{2,}"

# Find all phone numbers and email IDs
phone_numbers = re.findall(phone_pattern, data)
email_ids = re.findall(email_pattern, data)

# Print the results
print("Phone Numbers:", phone_numbers)
print("Email IDs:", email_ids)
```
