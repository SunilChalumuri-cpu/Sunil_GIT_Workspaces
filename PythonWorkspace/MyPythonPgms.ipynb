```python
def fact(n):
    if n==0 or n==1:
        return 1
    return n*fact(n-1)
print("Calling recursive function to find factorial of 5!=",fact(5))
```

```python
class homeToilet:
#Constructor of the class    
    def __init__(self,toilet):
        self.toilet = toilet
#Method of the class        
    def provideToilet(self):
        return self.toilet
#Creating a child class that inherits from the parent class homeToilet    
class myHome(homeToilet):
#Overriding the constructor of the parent class    
    def __init__(self,room,hall,kitchen):
        super().__init__("Common-toilet")
        self.room = room
        self.hall = hall
        self.kitchen = kitchen 
#Overriding the provideToilet method of the parent class               
    def provideToilet(self):
        return "Bedroom-toilet" 
#Method of the child class to permit the guest to stay in the home and use the toilet and kitchen facilities.       
    def permit(self,guest):
        try:
            if(guest == 'Sunil'):
                print("User:",guest,"can stay in",self.room,"and can also use",self.provideToilet())
            elif(guest == "Rama"):
                print("User:",guest,"can stay in",self.room,"and can also use",self.kitchen,",",self.provideToilet())
            else:
                print("User:",guest,"can stay in",self.hall,"and can also use",super().provideToilet())
        except Exception as e:
            print(e)
mh = myHome("Bedroom","Hall","Kitchen")
mh.permit(input("enter the Guest name"))
```

```python
class euron:
#Constructor of the class    
    def __init__(self, name, course, mentor, price):
        self.name = name
        self.course = course
        self.mentor = mentor
        self.price = price
#This method is to return the string representation of the object        
    def __str__(self):
        return f"Euron({self.name},{self.course},{self.mentor},{self.price})"
eu_obj1 = euron("sunil","python class","sudhanshu",4892)
print(eu_obj1)
```
